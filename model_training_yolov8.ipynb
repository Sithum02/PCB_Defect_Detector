```python
from ultralytics import YOLO

# Load the model
model = YOLO('yolov8m.pt')  # Use 'yolov8n.pt' for Nano, 'yolov8m.pt' for Medium, etc.

# Train the model
model.train(data='/content/dataset/pcb-defect-dataset/data.yml',  # Path to your dataset YAML file
            epochs=20,  # Number of epochs
            batch=16,   # Adjust based on GPU memory
            imgsz=640)  # Image size
```

```python
model.save('yolo_model.pt')
```
